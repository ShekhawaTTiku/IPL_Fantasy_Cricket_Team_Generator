## Concat all bbb data from different leagues

```python
import os

# Path to the folder containing the CSV files
folder_path = r'D:\all_league\all_match_files'

# List to store individual DataFrames
dataframes = []

# Iterate through all files in the folder
for file_name in os.listdir(folder_path):
    if file_name.endswith('_all_matches.csv'):
        # Extract league name from the file name
        league_name = file_name.split('_all_matches.csv')[0]

        # Read the CSV file
        file_path = os.path.join(folder_path, file_name)
        df = pd.read_csv(file_path, low_memory=False)

        # Add the league column
        df['league'] = league_name.upper()

        # Append the DataFrame to the list
        dataframes.append(df)

# Concatenate all DataFrames into one
combined_df = pd.concat(dataframes, ignore_index=True)

# Drop duplicate rows based on 'match_id', 'innings', and 'ball' columns
combined_df = combined_df.drop_duplicates(subset=['match_id', 'innings', 'ball'], keep='first')
```

```python
combined_df.sort_values(by=['league', 'start_date', 'match_id', 'innings', 'ball'], inplace=True)
```

## Concat all playing xi data from all leagues

```python
# Path to the folder containing the playing xi CSV files
folder_path = r'D:\all_league\playing_xi'

# List to store individual DataFrames
dataframes = []

# Iterate through all files in the folder
for file_name in os.listdir(folder_path):
    if file_name.endswith('_playing_xi.csv'):
        # Extract league name from the file name
        league_name = file_name.split('_playing_xi.csv')[0]

        # Read the CSV file
        file_path = os.path.join(folder_path, file_name)
        df = pd.read_csv(file_path, low_memory=False)

        # Add the league column
        df['league'] = league_name.upper()

        # Append the DataFrame to the list
        dataframes.append(df)

# Concatenate all DataFrames into one
combined_df = pd.concat(dataframes, ignore_index=True)

combined_df.drop(columns=['Unnamed: 0'], inplace=True)

# Save the combined DataFrame to a new CSV file
output_path = r'D:\all_league\all_league_playing_xi.csv'
combined_df.to_csv(output_path, index=False)
print(f"Combined file saved to: {output_path}")
```
