```python
import pandas as pd

def sort_csv_by_date(input_file, output_file):
    # Read the CSV file
    df = pd.read_csv(input_file, parse_dates=['date'])

    # Sort the dataframe by the 'date' column
    df_sorted = df.sort_values(by='date')

    # Save the sorted dataframe to a new CSV file
    df_sorted.to_csv(output_file, index=False)

    print(f"Sorted CSV saved as: {output_file}")

# Example usage
input_csv = "/content/ganjoo_ipl_bbb.csv"  # Replace with your actual file path
output_csv = "sorted_output.csv"
sort_csv_by_date(input_csv, output_csv)
```

```python
import pandas as pd

def process_ball_by_ball_data(input_file, output_file):
    # Read the CSV file
    df = pd.read_csv(input_file, parse_dates=['date'])

    # Group by date and match, then by bowler
    grouped = df.groupby(['date', 'p_match', 'bowl'])

    # Aggregate statistics
    summary = grouped.agg(
        overs=('ball', lambda x: len(x) // 6),
        balls=('ball', 'count'),
        wickets=('out', 'sum'),
        maiden_overs=('bowlruns', lambda x: sum(x.groupby(x.index // 6).sum() == 0)),
        economy=('bowlruns', lambda x: x.sum() / (len(x) / 6)),
        lbw_or_bowled=('dismissal', lambda x: sum(x.isin(['lbw', 'bowled'])))
    ).reset_index()

    # Sort by date, then match
    summary = summary.sort_values(by=['date', 'p_match'])

    # Save the processed data
    summary.to_csv(output_file, index=False)

    print(f"Processed match-by-match data saved as: {output_file}")

# Example usage
input_csv = "/content/ganjoo_ipl_sorted.csv"  # Replace with your actual file path
output_csv = "match_by_match.csv"
process_ball_by_ball_data(input_csv, output_csv)
```

```python
import pandas as pd

def process_ball_by_ball_data(input_file, output_file):
    # Read CSV file
    df = pd.read_csv(input_file, parse_dates=['date'])

    # Extract over number (over = ball // 6 + 1)
    df['over_number'] = df.groupby(['date', 'p_match', 'bowl']).cumcount() // 6 + 1

    # Group by date, match, and bowler
    grouped = df.groupby(['date', 'p_match', 'bowl'])

    # Aggregate statistics
    summary = grouped.agg(
        overs=('over_number', 'nunique'),  # More accurate overs calculation
        balls=('ball', 'count'),
        wickets=('out', 'sum'),
        runs=('bowlruns', 'sum'),
        maiden_overs=('over_number', lambda x: (df.loc[x.index].groupby(x)['bowlruns'].sum() == 0).sum()),  # Correct maiden over calc
        economy=('bowlruns', lambda x: x.sum() / (len(x) / 6) if len(x) > 0 else 0),
        lbw_or_bowled=('dismissal', lambda x: sum(x.isin(['lbw', 'bowled'])))
    ).reset_index()

    # Sort by date and match
    summary = summary.sort_values(by=['date', 'p_match'])

    # Save processed data
    summary.to_csv(output_file, index=False)
    print(f"Processed match-by-match data saved as: {output_file}")

# Example usage
input_csv = "/content/ganjoo_ipl_sorted.csv"  # Replace with actual file path
output_csv = "match_by_match_optimized.csv"
process_ball_by_ball_data(input_csv, output_csv)
```
