```python
import pandas as pd
import numpy as np
```

```python
# USE THIS WHEN RUNNING ON COLAB

import kagglehub

# Download latest version
path = kagglehub.dataset_download("rajatpatra/ig-bbb")

print("Path to dataset files:", path)
```

```python
df = pd.read_csv(path + "/t20_bbb_sorted.csv")
```

```python
df.shape
```

# Cleaning

```python
# Removing Unimportant Columns

df = df.drop(columns=['outcome','cur_bat_runs',
       'cur_bat_bf', 'cur_bowl_ovr', 'cur_bowl_wkts', 'cur_bowl_runs',
       'inns_runs', 'inns_wkts', 'inns_balls', 'inns_runs_rem',
       'inns_balls_rem', 'inns_rr', 'inns_rrr', 'target', 'max_balls'])
```

```python
# Keeping only thoose matches(or rows) which are played in major countries

countries_to_keep = ['South Africa', 'England', 'Australia', 'Bangladesh', 'India',
                     'Sri Lanka', 'West Indies', 'Pakistan', 'Scotland',
                     'Ireland', 'United Arab Emirates', 'New Zealand', 'United States of America']

df = df[df['country'].isin(countries_to_keep)]
```

```python
list(df['competition'].unique()) # all unique leagues left
```

```python
# Standardizing the Names of the League
# Define the dictionary with the replacements
dict_replacements = {
    'AUS v SL': 'T20I',
    'BBL 2023': 'BBL',
    'Ram Slam T20 Challenge': 'CSA T20',
    'Ram Slam T20': 'CSA T20',
    'CSA T20 Challenge': 'CSA T20',
    "Men's Hundred": "Men's 100",
    'Syed Mushtaq Ali Trophy': 'SMAT',
    'SMA TROPHY FINAL': 'SMAT',
    'SMA TROPHY': 'SMAT',
    'CPL 2023': 'CPL',
    'CPL 2024': 'CPL',
    'Bangladesh Premier League': 'BPL',
    'SLC Twenty-20 Tournament':'SLC T20'
}

# Use the replace method to update the 'competition' column
df['competition'] = df['competition'].replace(dict_replacements)
```

```python
# Keeping data from only the major leagues

major_leagues = [ 'T20I', 'BBL', 'SA20', 'T20 Blast', "Men's 100", 'IPL', 'SMAT', 'LPL', 'CPL', 'BPL', 'PSL', 'Super Smash', 'ILT20', 'MLC', 'CSA T20', 'MSL']
df = df[df['competition'].isin(major_leagues)]
```
