```python
import pandas as pd
import numpy as np
```

```python
venue_dict = {
    'Feroz Shah Kotla': 'Arun Jaitley Stadium',
    'Sardar Patel Stadium': 'Narendra Modi Stadium',
    'Dr. Y.S. Rajasekhara Reddy ACA-VDCA Cricket Stadium': 'Andhra Cricket Association-Visakhapatnam District Cricket Association Stadium',
    'Punjab Cricket Association Stadium':'Punjab Cricket Association IS Bindra Stadium',
    'Zayed Cricket Stadium': 'Sheikh Zayed Stadium',
    'Subrata Roy Sahara Stadium' : 'Maharashtra Cricket Association Stadium'
}
```

# Batting Data

```python
ganjoo_df_bat = pd.read_csv(r"C:\Users\HP\jupyter notebook files\FIFS_IPL\data\mbm\ganjoo_ipl_mbm_bat_final.csv")
csheet_df_bat = pd.read_csv(r"C:\Users\HP\jupyter notebook files\FIFS_IPL\data\mbm\cricsheet_ipl_mbm_bat_final.csv")
```

```python
min(ganjoo_df_bat['date']), max(ganjoo_df_bat['date']), min(csheet_df_bat['date']), max(csheet_df_bat['date'])
```

```python
combined_df_bat = pd.concat([ganjoo_df_bat, csheet_df_bat], ignore_index=True)
```

```python
combined_df_bat['date'] = pd.to_datetime(combined_df_bat['date'])
```

```python
max(combined_df_bat['date'])
```

```python
min(combined_df_bat['date'])
```

```python
combined_df_bat.info()
```

```python
combined_df_bat.to_csv('C:\\Users\\HP\\jupyter notebook files\\FIFS_IPL\\data\\mbm\\combined_ipl_mbm_bat.csv', index=False)
```

# Bowling Data

```python
ganjoo_df_bowl = pd.read_csv(r"C:\Users\HP\jupyter notebook files\FIFS_IPL\data\mbm\ganjoo_ipl_mbm_bowl_final.csv")
csheet_df_bowl = pd.read_csv(r"C:\Users\HP\jupyter notebook files\FIFS_IPL\data\mbm\cricsheet_ipl_mbm_bowl_final.csv")
```

```python
ganjoo_df_bowl['date'] = pd.to_datetime(ganjoo_df_bowl['date'], format='%d-%m-%Y')
csheet_df_bowl['date'] = pd.to_datetime(csheet_df_bowl['date'])
```

```python
min(ganjoo_df_bowl['date']), max(ganjoo_df_bowl['date'])
```

```python
min(csheet_df_bowl['date']), max(csheet_df_bowl['date'])
```

```python
ganjoo_df_bowl
```

```python
ganjoo_df_bowl['overs'] = ganjoo_df_bowl['overs'].replace('DNB', np.nan)
ganjoo_df_bowl['overs'] = ganjoo_df_bowl['overs'].replace(np.nan, 0)
ganjoo_df_bowl['overs']  = ganjoo_df_bowl['overs'].astype(float)
csheet_df_bowl['overs'] = csheet_df_bowl['overs'].replace('DNB', np.nan)
csheet_df_bowl['overs'] = csheet_df_bowl['overs'].replace(np.nan, 0)
csheet_df_bowl['overs'] = csheet_df_bowl['overs'].astype(float)
```

```python
combined_df_bowl = pd.concat([ganjoo_df_bowl, csheet_df_bowl], ignore_index=True)
```

```python
combined_df_bowl
```

```python
max(combined_df_bowl['date'])
```

```python
min(combined_df_bowl['date'])
```

```python
combined_df_bowl.info()
```

```python
combined_df_bowl['venue'] = combined_df_bowl['venue'].str.split(',').str[0]
combined_df_bowl['venue'] = combined_df_bowl['venue'].replace(venue_dict)
```

```python
combined_df_bowl.to_csv('C:\\Users\\HP\\jupyter notebook files\\FIFS_IPL\\data\\mbm\\combined_ipl_mbm_bowl.csv', index=False)
```
